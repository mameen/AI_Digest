# **Tools**

```python
# Cell 5: Calendar Tool
from google.adk.tools import FunctionTool
import json
from pathlib import Path

def read_calendar(date: str = "today"):
    """Read the user's calendar for a given date."""
    calendar_path = Path("data/calendar.json")
    if not calendar_path.exists():
        return {"error": "Calendar file not found", "tasks": []}
    
    with open(calendar_path, "r") as f:
        data = json.load(f)
    
    return {
        "date": data.get("date", date),
        "task_count": len(data.get("tasks", [])),
        "tasks": data.get("tasks", [])
    }

def prioritize_tasks(tasks_json: str):
    """Prioritize tasks by priority level, then by duration (shorter first)."""
    tasks = json.loads(tasks_json)
    priority_order = {"high": 0, "medium": 1, "low": 2}
    
    sorted_tasks = sorted(
        tasks,
        key=lambda t: (priority_order.get(t.get("priority", "low"), 2), t.get("duration_minutes", 999))
    )
    return sorted_tasks

calendar_tool = FunctionTool(read_calendar)
prioritize_tool = FunctionTool(prioritize_tasks)

print("✅ Calendar tools registered")
```

```python
# Cell 6: Email & Reminder Tools
def draft_email(recipient: str, subject: str, body: str, tone: str = "professional"):
    """Draft an email. Requires approval before sending."""
    return {
        "recipient": recipient,
        "subject": subject,
        "body": body,
        "tone": tone,
        "status": "DRAFT",
        "send_approved": False,
        "message": "Email drafted. Awaiting human approval before sending."
    }

def create_reminder(task_title: str, time: str, notes: str = ""):
    """Create a reminder for a task."""
    return {
        "task": task_title,
        "reminder_time": time,
        "notes": notes,
        "status": "REMINDER_SET"
    }

email_draft_tool = FunctionTool(draft_email)
reminder_tool = FunctionTool(create_reminder)

print("✅ Email & reminder tools registered")
```

```python
# Cell 7: PII Redaction Security Tool
import re

def redact_pii(text: str):
    """Redact PII (emails, phones, names) from text."""
    original = text
    redactions = []
    
    # Redact emails
    email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
    emails = re.findall(email_pattern, text)
    for i, email in enumerate(emails):
        placeholder = f"[EMAIL_{i+1}]"
        text = text.replace(email, placeholder)
        redactions.append({"type": "email", "original": email, "replacement": placeholder})
    
    # Redact phones
    phone_pattern = r'\b(?:\+?1[-.\s]?)?\(?\d{3}\)?[-.\s]?\d{3}[-.\s]?\d{4}\b'
    phones = re.findall(phone_pattern, text)
    for i, phone in enumerate(phones):
        placeholder = f"[PHONE_{i+1}]"
        text = text.replace(phone, placeholder)
        redactions.append({"type": "phone", "original": phone, "replacement": placeholder})
    
    # Redact names (First Last)
    name_pattern = r'\b([A-Z][a-z]+ [A-Z][a-z]+)\b'
    names = re.findall(name_pattern, text)
    for i, name in enumerate(names):
        placeholder = f"[PERSON_{i+1}]"
        text = text.replace(name, placeholder, 1)
        redactions.append({"type": "name", "original": name, "replacement": placeholder})
    
    return {
        "original_text": original,
        "sanitized_text": text,
        "pii_redacted": len(redactions) > 0,
        "redaction_count": len(redactions),
        "redactions": redactions
    }

def sanitize_for_logging(data: dict):
    """Sanitize a dict for logging."""
    text = json.dumps(data)
    result = redact_pii(text)
    return {
        "safe_for_logging": True,
        "sanitized_json": result["sanitized_text"],
        "redactions_applied": result["redaction_count"]
    }

pii_redact_tool = FunctionTool(redact_pii)
sanitize_log_tool = FunctionTool(sanitize_for_logging)

print("✅ PII redaction tools registered")
```

```python
# Cell 8: Human-in-the-Loop Approval Tool
from google.adk.tools import FunctionTool, ToolContext

def request_send_approval(draft_id: str, recipient: str, subject: str, preview: str, tool_context: ToolContext):
    """
    Request human approval before sending an email.
    PAUSES execution and waits for user confirmation.
    """
    tool_confirmation = tool_context.tool_confirmation
    
    if not tool_confirmation:
        tool_context.request_confirmation(
            hint=(
                f"📧 SEND APPROVAL REQUIRED\n\n"
                f"Draft ID: {draft_id}\n"
                f"To: {recipient}\n"
                f"Subject: {subject}\n\n"
                f"Preview:\n{preview}\n\n"
                f"Please approve (yes/no) or provide feedback."
            ),
            payload={"approved": False, "feedback": ""}
        )
        return {
            "status": "AWAITING_APPROVAL",
            "message": "Human confirmation required before sending. Please respond with approval."
        }
    
    approved = tool_confirmation.payload.get("approved", False)
    feedback = tool_confirmation.payload.get("feedback", "")
    
    if approved:
        return {
            "status": "APPROVED",
            "draft_id": draft_id,
            "message": "Send action approved by human. Proceeding...",
            "feedback": feedback
        }
    else:
        return {
            "status": "REJECTED",
            "draft_id": draft_id,
            "message": "Send action rejected by human. Draft cancelled.",
            "feedback": feedback
        }

approval_tool = FunctionTool(request_send_approval, require_confirmation=True)

print("✅ Approval tool registered")
```

```python
# Cell 10: Security Guardrail Agent
import os
from google.adk.agents import LlmAgent

GOOGLE_API_KEY = os.environ.get("GOOGLE_API_KEY", "")
if not GOOGLE_API_KEY:
    raise ValueError("GOOGLE_API_KEY not set! Set it in Cell 2.")

def security_check(text: str):
    """Run security check on text."""
    redaction_result = redact_pii(text)
    
    sensitive_keywords = ["password", "ssn", "credit card", "secret key", "api key"]
    lower_text = text.lower()
    flagged = [kw for kw in sensitive_keywords if kw in lower_text]
    
    return {
        "safe_to_process": len(flagged) == 0 and not redaction_result.get("pii_redacted", False),
        "pii_redacted": redaction_result.get("pii_redacted", False),
        "redaction_count": redaction_result.get("redaction_count", 0),
        "sanitized_text": redaction_result.get("sanitized_text", text),
        "flagged_keywords": flagged,
        "recommendation": "BLOCK" if flagged else ("REDACT" if redaction_result.get("pii_redacted") else "PASS")
    }

security_tool = FunctionTool(security_check)

security_guardrail = LlmAgent(
     model="gemini-2.5-flash",
    name="SecurityGuardrail",
    description="Screens all inputs/outputs for PII and sensitive content before processing or logging.",
    instruction=(
        "You are the Security Guardrail. Your job is to:\n"
        "1. Scan the user's input for PII (emails, phone numbers, names)\n"
        "2. Redact any detected PII using the security_check tool\n"
        "3. Flag sensitive keywords (passwords, API keys, SSNs, etc.)\n"
        "4. Return a security assessment: PASS, REDACT, or BLOCK\n"
        "5. If BLOCK: explain why and refuse to process\n"
        "6. If REDACT: provide the sanitized version for safe processing\n"
        "7. Save the sanitized input to state key 'sanitized_input' for the next agent\n\n"
        "Always be thorough but fast. Security is non-negotiable."
    ),
    tools=[security_tool, pii_redact_tool, sanitize_log_tool],
    output_key="security_report"
)

print("✅ Security Guardrail Agent created")
```

# **Root Agent and Runner**

```python
# # Cell 13: Root Sequential Agent
# from google.adk.agents import SequentialAgent

# Cell 13: Root SequentialAgent — FIXED with before_callback to pass context
from google.adk.agents import SequentialAgent

# BEFORE_CALLBACK: Copies original user message to state so all sub-agents can access it
def before_callback(callback_context):
    """Save original user message to state before any agent runs."""
    if callback_context.user_content and callback_context.user_content.parts:
        text = callback_context.user_content.parts[0].text
        callback_context.state["original_user_message"] = text
    return None  # Continue normally

root_agent = SequentialAgent(
    name="DailyConcierge",
    description="Your personal daily assistant. Plans your day, drafts emails, sets reminders, and keeps your data safe with PII redaction.",
    sub_agents=[
        security_guardrail,
        planner_agent,
        executor_agent,
    ],
    before_agent_callback=before_callback  # <-- CRITICAL: passes context to all agents
)

print("✅ Root SequentialAgent created")
print(f"Sub-agents: {[a.name for a in root_agent.sub_agents]}")
```

```python
# Cell 14: Run the Multi-Agent System — FIXED session attribute
import asyncio
from google.adk.runners import Runner
from google.adk.sessions import InMemorySessionService
from google.adk.artifacts import InMemoryArtifactService
from google.genai import types

APP_NAME = "daily_concierge"
USER_ID = "user_001"
SESSION_ID = "session_001"

async def run_concierge():
    session_service = InMemorySessionService()
    artifact_service = InMemoryArtifactService()
    
    session = await session_service.create_session(
        app_name=APP_NAME,
        user_id=USER_ID,
        session_id=SESSION_ID
    )
    
    # FIXED: Use correct attribute name (just 'id' or print the whole object)
    print(f"✅ Session created: {SESSION_ID}")  # Use the ID we passed in
    
    runner = Runner(
        agent=root_agent,
        app_name=APP_NAME,
        session_service=session_service,
        artifact_service=artifact_service,
    )
    
    user_message_text = (
        "Plan my day for July 7th. I need to send an email to alice.smith@partner.com "
        "about the contract review, and remind me about the team standup. "
        "Also look up the latest news on AI agents."
    )

    user_message = types.Content(
        role="user",
        parts=[types.Part(text=user_message_text)]
    )
    
    print("=" * 60)
    print("🤖 DAILY CONCIERGE — MULTI-AGENT SYSTEM")
    print("=" * 60)
    print(f"\n👤 User: {user_message_text}\n")
    print("-" * 60)
    
    print("\n📋 AGENT OUTPUT:\n")
    event_count = 0
    
    try:
        async for event in runner.run_async(
            user_id=USER_ID,
            session_id=SESSION_ID,
            new_message=user_message
        ):
            event_count += 1
            
            if event.content and event.content.parts:
                for part in event.content.parts:
                    if part.text:
                        text = part.text.strip()
                        print(f"  {text[:500]}")
        
        if event_count == 0:
            print("\n  ⚠️ No events received")
            
    except Exception as e:
        print(f"\n{'='*60}")
        print(f"❌ ERROR: {type(e).__name__}")
        print(f"{'='*60}")
        print(f"Full error: {str(e)}")
        if "429" in str(e):
            print("\n  → Rate limit hit. Wait 60 seconds and retry.")
    
    print(f"\n{'='*60}")
    print("✅ Execution complete!")
    print(f"{'='*60}")

await run_concierge()
```
